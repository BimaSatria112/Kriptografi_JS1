## 1. Persiapan

Pemetaan huruf yang digunakan adalah **A = 0, B = 1, ..., Z = 25**.

Untuk Caesar Cipher:
- Enkripsi: `C = (P + k) mod 26`
- Dekripsi: `P = (C - k) mod 26`

Huruf diproses tanpa membedakan huruf besar/kecil, sedangkan spasi dipertahankan.


```python
# Fungsi dasar Caesar Cipher

ALPHABET = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

def caesar_encrypt(text, key):
    result = ""
    for char in text.upper():
        if char in ALPHABET:
            index = ALPHABET.index(char)
            result += ALPHABET[(index + key) % 26]
        else:
            result += char
    return result

def caesar_decrypt(text, key):
    return caesar_encrypt(text, -key)
```

```python
def caesar_encrypt(pesan, key):
    hasil = ""

    for karakter in pesan:
        if karakter.isalpha():
            basis = ord("A") if karakter.isupper() else ord("a")
            hasil += chr((ord(karakter) - basis + key) % 26 + basis)
        else:
            hasil += karakter

    return hasil


def caesar_decrypt(ciphertext, key):
    return caesar_encrypt(ciphertext, -key)


pesan = input("Masukkan pesan: ")
key = int(input("Masukkan key (angka): "))

ciphertext = caesar_encrypt(pesan, key)
decrypted = caesar_decrypt(ciphertext, key)

print("\nHasil")
print("Plaintext  :", pesan.upper())
print("Ciphertext :", ciphertext)
print("Dekripsi   :", decrypted)
```

# MONOALPHABETICAL CIPHER

Pada bagian ini digunakan tabel substitusi tetap. Berdasarkan contoh pada jobsheet:
- A → Q
- B → W
- C → E
- D → R
- E → T
- T → Z

Pola tersebut sesuai dengan alfabet substitusi berbasis urutan keyboard QWERTY:

`ABCDEFGHIJKLMNOPQRSTUVWXYZ` → `QWERTYUIOPASDFGHJKLZXCVBNM`

Setiap huruf plaintext selalu memiliki pasangan huruf ciphertext yang tetap.


```python
# Tabel substitusi Monoalphabetical Cipher

ALPHABET = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
SUBSTITUTION = "QWERTYUIOPASDFGHJKLZXCVBNM"

encrypt_map = dict(zip(ALPHABET, SUBSTITUTION))
decrypt_map = dict(zip(SUBSTITUTION, ALPHABET))

print("Plaintext :", ALPHABET)
print("Ciphertext:", SUBSTITUTION)
```

```python
def mono_encrypt(text):
    result = ""
    for char in text.upper():
        if char in encrypt_map:
            result += encrypt_map[char]
        else:
            result += char
    return result

def mono_decrypt(text):
    result = ""
    for char in text.upper():
        if char in decrypt_map:
            result += decrypt_map[char]
        else:
            result += char
    return result
```

## 1. Substitusi Kata
Gunakan tabel substitusi untuk mengenkripsi **DATA**.


```python
plaintext = "DATA"
ciphertext = mono_encrypt(plaintext)

print("Plaintext :", plaintext)
print("Ciphertext:", ciphertext)
```

# VIGENÈRE CIPHER

Vigenère merupakan pengembangan Caesar Cipher. Key berupa kata atau rangkaian huruf, sehingga nilai pergeseran dapat berubah pada setiap posisi.

Rumus:
- Enkripsi: `C = (P + K) mod 26`
- Dekripsi: `P = (C - K) mod 26`

Spasi tidak dihitung sebagai posisi key sehingga key hanya digunakan pada huruf.


```python
def vigenere_encrypt(text, key):
    text = text.upper()
    key = key.upper()
    result = ""
    key_index = 0

    for char in text:
        if char in ALPHABET:
            p = ALPHABET.index(char)
            k = ALPHABET.index(key[key_index % len(key)])
            result += ALPHABET[(p + k) % 26]
            key_index += 1
        else:
            result += char

    return result

def vigenere_decrypt(text, key):
    text = text.upper()
    key = key.upper()
    result = ""
    key_index = 0

    for char in text:
        if char in ALPHABET:
            c = ALPHABET.index(char)
            k = ALPHABET.index(key[key_index % len(key)])
            result += ALPHABET[(c - k) % 26]
            key_index += 1
        else:
            result += char

    return result
```

## Evaluasi 2 — Caesar Cipher dengan Huruf, Spasi, dan Angka

Pesan pada sistem dapat mengandung huruf, spasi, dan angka. Program Caesar Cipher harus 
**tidak mengubah spasi dan angka**, sedangkan huruf tetap digeser berdasarkan key.

Keputusan ini dibuat agar informasi non-huruf pada pesan tetap sama setelah proses enkripsi. 
Dengan demikian, hanya karakter alfabet `A-Z` atau `a-z` yang mengalami pergeseran.

```python
# Evaluasi 2: Caesar Cipher yang mempertahankan spasi dan angka

def caesar_evaluasi_2(text, key):
    hasil = ""

    for karakter in text:
        if karakter.isalpha() and karakter.upper() in ALPHABET:
            basis = ord("A") if karakter.isupper() else ord("a")
            hasil += chr((ord(karakter) - basis + key) % 26 + basis)
        else:
            # Spasi, angka, dan karakter non-alfabet lainnya tidak diubah
            hasil += karakter

    return hasil


pesan = "DATA AMAN 2026"
key = 5

ciphertext = caesar_evaluasi_2(pesan, key)
decrypted = caesar_evaluasi_2(ciphertext, -key)

print("=== HASIL EVALUASI 2 ===")
print("Plaintext  :", pesan)
print("Key        :", key)
print("Ciphertext :", ciphertext)
print("Dekripsi   :", decrypted)
```

### Penjelasan Evaluasi 2

Pada fungsi `caesar_evaluasi_2()`, kondisi `karakter.isalpha()` digunakan untuk memeriksa 
karakter alfabet. Jika karakter merupakan huruf A-Z atau a-z, karakter tersebut digeser 
menggunakan rumus Caesar Cipher.

Jika karakter bukan huruf, program langsung memasukkannya ke hasil tanpa perubahan. 
Karena itu, **spasi tetap menjadi spasi dan angka tetap menjadi angka**.

Pada contoh `DATA AMAN 2026` dengan key `5`, bagian `DATA AMAN` mengalami pergeseran, 
sedangkan spasi dan `2026` tetap sama.

## Evaluasi 4 — Membuat Alfabet Rahasia Sendiri

Sebuah kelompok ingin membuat alfabet rahasia sendiri. Alfabet tersebut harus memiliki 
26 pasangan huruf dan tidak menggunakan pola pergeseran Caesar.

Pada evaluasi ini digunakan susunan alfabet acak berikut:

`MNBVCXZLKJHGFDSAPOIUYTREWQ`

Program menggunakan dictionary untuk menyimpan pasangan huruf. Dictionary enkripsi 
digunakan untuk mencari pengganti dari plaintext, sedangkan dictionary dekripsi dibuat 
dengan membalik pasangan tersebut.

```python
# Evaluasi 4: Alfabet rahasia sendiri tanpa pola Caesar

RAHASIA_ALPHABET = "MNBVCXZLKJHGFDSAPOIUYTREWQ"

# Membuat pasangan enkripsi dan dekripsi
rahasia_encrypt_map = dict(zip(ALPHABET, RAHASIA_ALPHABET))
rahasia_decrypt_map = dict(zip(RAHASIA_ALPHABET, ALPHABET))


def rahasia_encrypt(text):
    hasil = ""

    for karakter in text.upper():
        if karakter in rahasia_encrypt_map:
            hasil += rahasia_encrypt_map[karakter]
        else:
            hasil += karakter

    return hasil


def rahasia_decrypt(text):
    hasil = ""

    for karakter in text.upper():
        if karakter in rahasia_decrypt_map:
            hasil += rahasia_decrypt_map[karakter]
        else:
            hasil += karakter

    return hasil


# Pemeriksaan bahwa alfabet rahasia merupakan permutasi 26 huruf
print("=== ALFABET RAHASIA ===")
print("Plaintext :", ALPHABET)
print("Ciphertext:", RAHASIA_ALPHABET)
print("Alfabet valid:", len(RAHASIA_ALPHABET) == 26 and
      len(set(RAHASIA_ALPHABET)) == 26 and
      set(RAHASIA_ALPHABET) == set(ALPHABET))

# Contoh enkripsi dan dekripsi
plaintext = "PESAN RAHASIA"
ciphertext = rahasia_encrypt(plaintext)
decrypted = rahasia_decrypt(ciphertext)

print("\n=== HASIL EVALUASI 4 ===")
print("Plaintext  :", plaintext)
print("Ciphertext :", ciphertext)
print("Dekripsi   :", decrypted)
print("Berhasil kembali ke plaintext:", decrypted == plaintext)
```

### Penjelasan Evaluasi 4

Program mengetahui pasangan setiap huruf melalui `rahasia_encrypt_map`. Fungsi `zip()` 
memasangkan alfabet asli dengan alfabet rahasia berdasarkan posisi yang sama.

Sebagai contoh, huruf pertama `A` dipasangkan dengan `M`, huruf kedua `B` dengan `N`, 
dan seterusnya. Untuk proses dekripsi, pasangan tersebut dibalik sehingga ciphertext 
dapat dicari kembali menjadi plaintext.

Berbeda dari Caesar Cipher, alfabet rahasia ini tidak dibuat dengan menggeser seluruh huruf 
menggunakan jumlah pergeseran yang sama. Keamanan metode tetap bergantung pada kerahasiaan 
tabel substitusinya.

### Menunjukkan Pengulangan Key

Pada Vigenère Cipher, key `DATA` akan digunakan berulang selama masih terdapat huruf 
pada plaintext. Karena spasi dan angka tidak diproses, karakter key hanya maju ketika 
program menemukan huruf.

`D A T A D A T A D A T A ...`

```python
# Menampilkan key yang benar-benar digunakan pada setiap huruf plaintext

def vigenere_key_sequence(text, key):
    key = key.upper()
    sequence = ""
    key_index = 0

    for karakter in text.upper():
        if karakter in ALPHABET:
            sequence += key[key_index % len(key)]
            key_index += 1
        else:
            sequence += karakter

    return sequence


print("Plaintext :", plaintext)
print("Key       :", vigenere_key_sequence(plaintext, key))
```
